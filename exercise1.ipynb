```python
"""
Q1. Write a function that takes a string and returns:
-	reversed string
-	string with every 2nd character
"""

text = "Hello World"
rev_text = text[::-1]
print(f"The reversed string is : {rev_text}")

# string with every second character

str2 = text[0::2]
print(f"The string with every second character is : {str2}")
```

```python
"""
•	Q2. Given a list of numbers, return a new list containing squares of even numbers
only using list comprehension.
"""

list_no = [10 , 20 , 25 , 30 , 35, 40 , 50 ]
sq_list = [i*i for i in list_no if i %2 == 0]
print(f"The list with squares of even numbers is : {sq_list}")
```
