```python
"""
Q1. Write a function that takes a string and returns:
-	reversed string
-	string with every 2nd character
"""

text = "Hello World"
rev_text = text[::-1]
print(f"The reversed string is : {rev_text}")

# string with every second character

str2 = text[0::2]
print(f"The string with every second character is : {str2}")
```

```python
"""
•	Q2. Given a list of numbers, return a new list containing squares of even numbers
only using list comprehension.
"""

list_no = [10 , 20 , 25 , 30 , 35, 40 , 50 ]
sq_list = [i*i for i in list_no if i %2 == 0]
print(f"The list with squares of even numbers is : {sq_list}")
```

```python
"""
•	Q3. Write a function that counts frequency of each character in a string using a dictionary.
"""
def freq_char(str1):
  dict = {}

  for char in str1:
     dict.update({char : str1.count(char)})
print(f"The frequency of each character in a string is : {dict}")

str1 = "appleapple"
freq_char(str1)
```

```python
# •	Q4. Write a function that takes a list and returns indices of all odd numbers using enumerate().

def ret_indx(list_no):
  new_list = []
  for i , element in enumerate(list_no):
    if element % 2 != 0:
      new_list.append(i)
  return new_list


list_no = [5 ,10 , 15 , 20 , 25 , 30 , 35, 40 , 45]
new_list = ret_indx(list_no)
print(f"The indices of all odd numbers in a list is : {new_list}")
```
