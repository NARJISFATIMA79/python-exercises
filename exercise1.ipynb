```python
"""
Q1. Write a function that takes a string and returns:
-	reversed string
-	string with every 2nd character
"""

text = "Hello World"
rev_text = text[::-1]
print(f"The reversed string is : {rev_text}")

# string with every second character

str2 = text[0::2]
print(f"The string with every second character is : {str2}")
```

```python
"""
•	Q2. Given a list of numbers, return a new list containing squares of even numbers
only using list comprehension.
"""

list_no = [10 , 20 , 25 , 30 , 35, 40 , 50 ]
sq_list = [i*i for i in list_no if i %2 == 0]
print(f"The list with squares of even numbers is : {sq_list}")
```

```python
"""
•	Q3. Write a function that counts frequency of each character in a string using a dictionary.
"""
def freq_char(str1):
  dict = {}

  for char in str1:
     dict.update({char : str1.count(char)})
print(f"The frequency of each character in a string is : {dict}")

str1 = "appleapple"
freq_char(str1)
```

```python
# •	Q4. Write a function that takes a list and returns indices of all odd numbers using enumerate().

def ret_indx(list_no):
  new_list = []
  for i , element in enumerate(list_no):
    if element % 2 != 0:
      new_list.append(i)
  return new_list


list_no = [5 ,10 , 15 , 20 , 25 , 30 , 35, 40 , 45]
new_list = ret_indx(list_no)
print(f"The indices of all odd numbers in a list is : {new_list}")
```

```python
# Q5. Write a function to rotate a list right by k positions using slicing.
def rotate_right(lst, k):
    k = k % len(lst)  # handles k > length
    return lst[-k:] + lst[:-k]

lst = [1,2,3,4,5]
k = 2

rotate_right(lst , k)
```

```python
# Q6. Write a function that takes two lists and returns a dictionary using zip().
def lists_to_dict(keys, values):
    return dict(zip(keys, values))

keys = [1 ,2, 3, 4, 5]
values = ["apple" , "mango" , "banana" , "peach" "strawberry"]

lists_to_dict(keys , values)


```

```python
"""" • Q7. Using iter() and next(), manually iterate through a list and print
elements one by one. Also, perform exception handling for the exception
StopIteration using a try-except block.
"""

#manual iteration
def manual_iteration(lst):
    it = iter(lst)
    while True:
        try:
            print(next(it))
        except StopIteration:
            break

lst = [1 ,2,3,4,5]
manual_iteration(lst)

```

```python
# Flatten nested list (list comprehension).

def flatten_list(nested):
    return [item for sublist in nested for item in sublist]

nested = [[1 , 2 ], [3 , 4,5 ]]

flatten_list(nested)
```

```python
# Q9 — Palindrome using slicing

def is_palindrome(s):
    return s == s[::-1]

str1 = "mam"
is_palindrome(str1)
```

```python
# Q10 — Second largest (without sorting)

def second_largest(lst):
    first = second = float('-inf')

    for num in lst:
        if num > first:
            second = first
            first = num
        elif first > num > second:
            second = num

    return second if second != float('-inf') else None

lst = [1 ,2 ,3, 4, 5]

second_largest(lst)
```
